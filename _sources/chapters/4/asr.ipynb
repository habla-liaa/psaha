# Reconocimiento Automático del Habla (ASR)

**Objetivo:** Pasar habla a texto (speech-to-text, stt).



## Formulación Probabilística

Dada una secuencia de observaciones \(O\) (onda acústica), queremos encontrar la secuencia de palabras \(W\) que la generó. Esto se puede expresar como:

$$
\hat{W} = \arg\max_{W} P(W|O) 
$$

Aplicando la regla de Bayes obtenemos:

$$
\hat{W} = \arg\max_{W} \frac{P(O|W) P(W)}{P(O)}
$$

El denominador \(P(O)\) no depende de \(W\), por lo que podemos ignorarlo. Así que nos quedamos con:

$$
\hat{W} = \arg\max_{W} P(O|W) P(W)
$$

Donde $P(O|W)$ está dado por el modelo acústico y $P(W)$ por el modelo de lenguaje.
