#  Sistemas Lineales e Invariantes en el Tiempo


## Sistema lineal

Definiendo un sistema lineal como aquel que cumple con la propiedad de superposición y de homogeneidad.

### Superposición
Se aplica una entrada $x_1$ y se obtiene una salida $y_1$ y luego se aplica una entrada $x_2$ y se obtiene una salida $y_2$, entonces si se aplica una entrada $x_1 + x_2$ se obtiene una salida $y_1 + y_2$.

$$ x_1 \rightarrow S \rightarrow y_1, \quad y_2 \rightarrow S \rightarrow y_2 $$
$$ x_1 + x_2 \rightarrow S \rightarrow y_1 + y_2 $$

Descripto con notación de operador lo podemos expresar como:

$$
y_1(t) = H \cdot x_1(t), \quad y_2(t) = H \cdot x_2(t)
$$

$$
y_1(t) + y_2(t) = H \cdot (x_1(t) + x_2(t))
$$

El resultado de un sistema a estimulos complejos puede expresarse como estímulos menos complejos.

### Homogeneidad
Si se aplica una entrada $\alpha x$ se obtiene una salida $\alpha y$.

$$ x \rightarrow S \rightarrow y $$

$$ \alpha x \rightarrow S \rightarrow \alpha y $$

Con notación de operador:

$$
y(t) = H \cdot x(t)
$$

$$
\alpha y(t) = \alpha H \cdot x(t) = H \cdot (\alpha x(t)) 
$$


## Sistema invariante en el tiempo

Un sistema es invariante en el tiempo si el desplazamiento temporal de la entrada produce un desplazamiento temporal en la salida, es decir el sistema no cambia su comportamiento en el tiempo.

$$ x(t) \rightarrow S \rightarrow y(t) $$
$$ x(t - t_0) \rightarrow S \rightarrow y(t - t_0) $$

Con notación de operador:

$$
x(t-t_0) = D_{t_0} \cdot x(t) 
$$

$$
y(t - t_0) = H \cdot x(t - t_0) = H \cdot D_{t_0} \cdot x(t) = D_{t_0} \cdot H \cdot x(t) = D_{t_0} \cdot y(t)
$$

## Caracterización de un sistema

Cualquier señal la podemos expresar como:

$$
\begin{align}
x[n] = \sum_{k=-\infty}^{\infty} x[n] \delta[k-n] & \, & x(t) = \int_{-\infty}^{\infty} x(t) \delta(t - \tau) d\tau \tag{1}
\end{align}
$$

Retomemos la idea de generar una salida compleja en base a combinaciones de señales más simples. En este caso, cualquier señal puede ser expresada como una combinación lineal de impulsos.

Por otro lado pensemos como representar al sistema identidad, el cual donde la señal de salida es igual a la señal de entrada. La forma trivial sería simplemente multiplicar la señal de entrada por 1.

$$
x[n] = 1 \cdot x[n] 
$$

Al mismo tiempo la expresión (1), también aplica como un sistema LTI y también representa la identidad. Podemos generalizar a caracterizar cualquier sistema LTI, extendiendo la expresión (1) pensando que una salida será la composición de entradas desplazadas y escaladas.

$$ y[n] = \alpha_0 x[n-0] + \alpha_1 x[n-1] + \alpha_2 x[n-2] + \cdots $$

Si pensamos estos coeficientes como una nueva señal $h[n]$, podemos expresar la salida como:

$$ y[n] = h[0]x[n-0] + h[1]x[n-1] + h[2]x[n-2] + \cdots $$

Y de forma general

$$ y[n] = \sum_{k=-\infty}^{\infty} h[k]x[n-k] $$

Esta expresión se conoce como la convolución entre las señales $x[n]$ y $h[n]$.

$$
y[n] = \left(h * x\right)[n]
$$

Por lo tanto el comportamiento de un sistema LTI queda caracterizado por la señal $h[n]$.

El análogo en el caso continuo es:

$$ y(t) =  \int_{-\infty}^{\infty} h(\tau) x(t - \tau) d\tau  =  \left( h * x \right)(t)\  $$

## Respuesta al impulso de un sistema LTI

Imaginemos tenemos acceso a un sistema LTI, al cual le podemos aplicar cualquier señal de entrada y medir la señal de salida. Sería interesante poder encontrar $h[n]$ y caracterizar el sistema en base a una señal de entrada especial. Si vemos las ecuación (1), la señal $\delta[n]$ cumple con la propiedad de ser el elemento neutro, es decir que al convolucionar cualquier señal con el impulso, la señal de salida es igual a la señal de entrada. En el caso de un sistema LTI caracterizado por una señal $h[n]$, aplicar un impulso genera la señal que caracteriza al sistema, esto le da a $h[n]$ el nombre de respuesta al impulso del sistema.



$$
\begin{align}
y[n] &= \left(h * \delta\right)[n] = h[n] &\quad y(t) &= \left( h * \delta \right)(t) = h(t)
\end{align}
$$


## Representación de un sistema LTI

Un sistema LTI puede ser representado de diferentes maneras, dependiendo del contexto. En el caso discreto, se puede representar como un sistema de ecuaciones en diferencias, mientras que en el caso continuo se puede representar como un sistema de ecuaciones diferenciales.

### Ecuación general de un sistema discreto LTI

La siguiente notación es comúnmente utilizada para representar un sistema LTI discreto, se establece que hay dos grupos de coeficientes, uno para la señal de entrada $b_k$ y otro para la señal de salida $a_k$. La ecuación general de un sistema LTI discreto es:

$$
\sum_{k=0}^{N} a_k y[n-k] = \sum_{k=0}^{M} b_k x[n-k]
$$

### Ecuación general de un sistema lineal de ecuaciones diferenciales

$$
\sum_{k=0}^{N} a_k \frac{d^k y(t)}{dt^k} = \sum_{k=0}^{M} b_k \frac{d^k x(t)}{dt^k}
$$

## Autofunciones de la convolución / sistema LTI



$$
\begin{align}
\left(h * x\right)[n] &= c x[n] \\
\left( h * x \right)(t) &= c x(t)
\end{align}
$$

Dado que todo sistema LTI puede verse como aplicar a una señal de entrada la convolución sobre la respuesta al impulso, podemos preguntarnos que señales de entrada se comportan como autofunciones, es decir el resultado de la convolución es una señal proporcional a la señal de entrada. La principal propiedad que deben tener las autofunciones es que deben transformar los desplazamientos de la variable independiente en una multiplicación por un escalar para poder extraerlo de la integral o sumatoria. La respuesta para el caso continuo está dada por la familia de funciones exponenciales complejas $e^{s t}$ donde $s$ es un número complejo y para el caso discreto por $z^{n}$ donde $z$ es un número complejo.  

$$
\begin{align}
x(t) &= e^{s t} \\
y(t) &= \int_{-\infty}^{\infty} h(\tau) x(t - \tau) d\tau = \int_{-\infty}^{\infty} h(\tau) e^{s (t - \tau)} d\tau \\
y(t) &= e^{s t} \int_{-\infty}^{\infty} h(\tau) e^{-s \tau} d\tau \\
y(t) &= e^{s t} H(s) \\
y(t) &= H(s) x(t)
\end{align}
$$

Para el caso discreto:

$$
\begin{align}
x[n] &= z^{n} \\
y[n] &= \sum_{k=-\infty}^{\infty} h[k] x[n-k] = \sum_{k=-\infty}^{\infty} h[k] z^{n-k} \\
y[n] &= z^{n} \sum_{k=-\infty}^{\infty} h[k] z^{-k} \\
y[n] &= z^{n} H(z) \\
y[n] &= H(z) x[n]
\end{align}
$$
