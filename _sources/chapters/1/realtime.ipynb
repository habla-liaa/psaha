```python
import numpy as np
import soundcard as sc
```

```python
# get a list of all speakers:
speakers = sc.all_speakers()
# get the current default speaker on your system:
default_speaker = sc.default_speaker()
# get a list of all microphones:
default_speaker

data = np.random.randn(48000*2, 2)
data = data/np.max(np.abs(data))
default_speaker.play(data, samplerate=48000)
```

```python
nframes = 1024
sr = 48000
n0 = 0
with default_speaker.player(samplerate=sr) as sp:
    for _ in range(100):
        data = np.sin(2*np.pi*440*np.arange(n0,n0+nframes)/sr)
        n0 += nframes
        sp.play(data)
```

```python
import time
def print_numbers():
    for i in range(1, 11):
        print(i)
        time.sleep(1)
```

```python
# run a thread that prints 1 to 10 with 1 second delay
# and then stops

from threading import Thread

t = Thread(target=print_numbers)
t.start()
```
