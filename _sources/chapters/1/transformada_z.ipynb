# Visualización de funciones sobre números complejos

$\displaystyle  f(z) = z$

```python
func = lambda z: 10*z

def func(z):
    return z

x = y = np.linspace(-5,5,100)
complexfunction(func, x,y)
plt.vlines(0,-2,2)
plt.hlines(0,-2,2);
plt.xlim(-2,2)
plt.ylim(-2,2);
```

$\displaystyle  f(z) = \left|z\right|$

```python
func = lambda z: abs(z)
x = y = np.linspace(-2,2,100)
complexfunction(func, x,y)
plt.vlines(0,-2,2)
plt.hlines(0,-2,2);
plt.xlim(-2,2)
plt.ylim(-2,2);
```

$\displaystyle f(z) = z^2$

```python
func = lambda z: z**2
x = y = np.linspace(-5,5,100)
complexfunction(func, x,y)
plt.vlines(0,-2,2)
plt.hlines(0,-2,2);
plt.xlim(-2,2)
plt.ylim(-2,2);
```

$\displaystyle  f(z) = z-1$

```python
func = lambda z: z-1
x = y = np.linspace(-10,10,200)
complexfunction(func, x,y)
plt.vlines(0,-2,2)
plt.hlines(0,-2,2);
plt.xlim(-2,2)
plt.ylim(-2,2);
```

$\displaystyle  f(z) = \frac{1}{z-z0}$

$\displaystyle  z0 = 0.6+i0.6$

```python
func = lambda z: 1/(z-0.6-0.6*1j)
x = y = np.linspace(-5,5,200)
complexfunction(func, x,y)
plt.vlines(0,-2,2)
plt.hlines(0,-2,2);
plt.xlim(-2,2)
plt.ylim(-2,2);
```
