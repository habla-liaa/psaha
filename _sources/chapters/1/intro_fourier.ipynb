# Introducción Fourier

La teoria de Fourier permite represenar señales expresadas en el dominio del tiempo como una composición de señales tipo senos y cosenos de distintas aplitudes, frecuencias y fases. Nos interesa expresar las operaciones que permiten computar la representación de Fourier de una señal (análisis), así como la forma de recuperar la señal original a partir de su representación en el dominio de Fourier (síntesis).

Distinguiremos entre varios tipos de represetaciones de Fourier según trabajemos en el dominio continuo o discreto, y según trabajemos con señales periódicas o no periódicas.

Las principales representaciones de Fourier son:
- Serie de Fourier
    - Señales continuas periódicas
    - Suma de infinitas componentes
    - Frecuencias discretas
- Transformada de Fourier
    - Señales continuas no periódicas
    - Integral de infinitas componentes
    - Frecuencias continuas
- Transformada de Fourier de tiempo discreto y frecuencia discreta (DFT)
    - Señales discretas periódicas
    - Suma de finitas componentes
    - Frecuencias discretas
- Transformada de Fourier de tiempo discreto (DTFT)
    - Señales discretas no periódicas
    - Integral de finitas componentes
    - Frecuencias continuas
  
## Serie de Fourier 
Representación de señales $L^2_{(a,b)}$ (funciones cuadrado integrables) como una suma infinita de senos y cosenos. Dado que se trabaja sobre un intevalo de tiempo finito $(a,b)$, la series de fourier sirve para representar señales periódicas de período $T=b-a$ como una suma infinita de senos y cosenos con frecuencias múltiplos de una frecuencia fundamental $w_0 = \frac{2 \pi}{T}$.

Síntesis:

$$
x(t) = \sum_{n=-\infty}^{\infty} c_n e^{i n \omega_0 t}
$$

Análisis:

$$
c_n = \frac{1}{T} \int_a^b x(t) e^{-i n \omega_0 t} dt
$$

## Transformada de Fourier

Representación de señales $L^2_{(-\infty,\infty)}$ (funciones cuadrado integrables) como una integral de senos y cosenos. La transformada de Fourier permite representar señales no periódicas como una suma infinita de senos y cosenos con frecuencias continuas.

Síntesis:
$$
x(t) = \frac{1}{2 \pi} \int_{-\infty}^{\infty} X(w) e^{i w t} dw
$$
Análisis:
$$
X(w) = \int_{-\infty}^{\infty} x(t) e^{-i w t} dt
$$

## Transformada de Fourier discreta (DFT)

Representación de señales discretas periódicas $x[n]$ con $N$ muestras como una suma finita de senos y cosenos en frecuencias dadas por múltiplos de $w_0 = \frac{2 \pi}{N}$. 

Síntesis:
$$
x[n] = \frac{1}{N} \sum_{k=0}^{N-1} X[k] e^{i w_0 k n} 
$$
Análisis:
$$
X[k] = \sum_{n=0}^{N-1} x[n] e^{-i w_0 k n}
$$

## Transformada de Fourier discreta de tiempo (DTFT)

Representación de señales discretas no periódicas $x[n]$ como una integral finita de senos y cosenos en frecuencias continuas.

Síntesis:
$$
x[n] = \frac{1}{2 \pi} \int_{-\pi}^{\pi} X(w) e^{i w n} dw
$$
Análisis:
$$
X(w) = \sum_{n=-\infty}^{\infty} x[n] e^{-i w n}
$$
