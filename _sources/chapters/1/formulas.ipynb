```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import fftconvolve
```

```python
# fft convolve of a ramp with dirichlet kernel
sr = 1000
T = 2
t = np.arange(0, int(sr*T))/sr-T/2
x = np.arange(0, int(sr*T))/sr-T/2
                                                                                                
# dirichlet kernel
def dirichlet_kernel(x,n):
    x = x.copy()
    mask = x==0
    x[mask] = 1
    d = np.sin( (n+1/2)*x ) / np.sin(x/2)
    d[mask] = n*2 + 1
    return d
```

```python
n = 100
d = dirichlet_kernel(t, n)
d = d/n /2/np.pi
y = fftconvolve(x, d, mode='same')
plt.plot(x)
plt.plot(y);
```
