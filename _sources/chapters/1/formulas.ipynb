## Sistemas de procesamiento

A una señal se le puede aplicar un proceso o transformación. Tradicionalmente pensamos en una función matemática que toma una señal $x$ como argumento y devuelve otra señal $y = f(x)$. Podemos generalizar esto a la idea de un sistema que toma una señal de entrada, la procesa y genera otra señal de salida.

img entrada -> proceso -> salida

Los sistemas más simples son aquellos que son lineales y no cambian el dominio de la señal. Por ejemplo,  tomar una señal de entrada y multiplicarla por una constante.

Como casos más generales podemos pensar sistemas que realizan procesos no lineales, cajas negras, sistemas que cambian el dominio o la dimensionalidad de la señal. Por ejemplo un sistema que toma habla y devuelve texto.
