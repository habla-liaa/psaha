# Series de Fourier

El origen histórico de la serie de Fourier está relacionado a la solución de ecuaciones diferenciales en derivadas parciales, como puedes ser la ecuación de ondas o del calor.

Una forma de derivar la serie de Fourier es estudiar las soluciones de problemas de osciladores acoplados y su caso límite obteniendo la ecuación de ondas en una cuerda. Los sistemas de osciladores acoplados pueden plantearse de forma matricial, y una técnica para encontrar las ecuaciones de movimiento es encontrar una base que permita diagonalizar la matriz de acoplamiento. Esta es la base de autovectores de la matriz. Esto permite obtener un sistema equivalente de osciladores independientes que generan los llamados modos normales. La solución general del sistema de osciladores acoplados es una combinación lineal de estos modos normales. Esto es un propiedad básica de alebra lineal que permite expresar cualquier vector como una combinación lineal de elementos de una base. Este propiedad se puede extender al caso de espacios vectoriales de infinita dimensión, es decir, el caso de un sistema continuo, como una cuerda. La solución general se puede expresar como una serie infinita de funciones ortogonales, que son las funciones propias del operador diferencial asociado al problema. En el caso de la ecuación de ondas, estas funciones propias son senos y cosenos, y la serie resultante es la serie de Fourier.

A continuación presentamos un derivación más formal de la serie de Fourier. La idea es representar una función periódica como una suma infinita de senos y cosenos (o exponenciales complejas), que son funciones ortogonales.

## Producto interno entre funciones

Consideremos un espacio de funciones  $L^2_{\left(a,b\right)}$, el cual es el espacio de funciones cuadrado integrables en el intervalo $\left(a,b\right)$, es decir, funciones $f$ tales que:

$$
\int_a^b |f(t)|^2 dt < \infty
$$

Este espacio es un espacio vectorial, y podemos definir un producto interno entre dos funciones $f$ y $g$ en $L^2_{\left(a,b\right)}$ como:

$$
\langle f, g \rangle = \int_a^b f(t) g^*(t) dt
$$

donde $g^*(t)$ es el conjugado complejo de $g(t)$. Este producto interno nos permite definir la norma de una función como:

$$
||f|| = \sqrt{\langle f, f \rangle} = \sqrt{\int_a^b |f(t)|^2 dt}
$$

Si el producto es nulo, es decir, $\langle f, g \rangle = 0$, entonces las funciones son ortogonales. 

### Aplicación a señales

Observar que la norma cuadrado de una función es equivalente a la energía de la señal en el intervalo $\left(a,b\right)$, es decir:

$$
E = ||f||^2 = \int_a^b |f(t)|^2 dt
$$

Por otro lado el producto interno es una medida de la correlación entre dos funciones:

$$
R_{fg} = \langle f, g \rangle = \int_a^b f(t) g^*(t) dt
$$

## Base de funciones ortogonales

Una base de funciones ortogonales es un conjunto de funciones $\{f_n(t)\}$ tales que:
1. Cada función $f_n(t)$ es ortogonal a las demás, es decir, $\langle f_n, f_m \rangle = 0$ para $n \neq m$.
2. Cualquier función $f(t)$ en el espacio $L^2_{\left(a,b\right)}$ se puede expresar como una combinación lineal de las funciones de la base, es decir:
$$
f(t) = \sum_{n=1}^{\infty} c_n f_n(t)
$$
donde $c_n = \langle f, f_n \rangle$ son los coeficientes.

## Base ortogonal para funciones periódicas (series de Fourier)

Consideremos funciones periódicas con periodo $T$, es decir, funciones que cumplen la propiedad:
$$
f(t + T) = f(t)
$$
Una base ortogonal para funciones periódicas es el conjunto de funciones 

$$\{e^{i \frac{2 \pi n}{T} t}\}$$

donde $n$ es un entero. Estas funciones son ortogonales entre sí en el intervalo $\left(0, T\right)$, es decir:

$$
\langle e^{i \frac{2 \pi n}{T} t}, e^{i \frac{2 \pi m}{T} t} \rangle = \int_0^T e^{i \frac{2 \pi n}{T} t} e^{-i \frac{2 \pi m}{T} t} dt = \int_0^T e^{i \frac{2 \pi (n - m)}{T} t} dt = T \delta_{nm}
$$

Por lo tanto, si $n \neq m$ el producto interno es cero, y si $n = m$ el producto interno es igual a $T$.

La integral de una exponencial compleja se puede resolver como: 
$$
\int_0^T e^{i \frac{2 \pi (n - m)}{T} t} dt = \left[ \frac{T}{i 2 \pi (n - m)} e^{i \frac{2 \pi (n - m)}{T} t} \right]_0^T = \frac{T}{i 2 \pi (n - m)} \left( e^{i 2 \pi (n - m)} - 1 \right) = 0
$$

Para $n \neq m$, ya que $e^{i 2 \pi (n - m)} = 1$. Por lo tanto, el producto interno entre funciones de diferentes frecuencias es cero, lo que implica que son ortogonales.
Para $n = m$, la integral se convierte en:
$$
\int_0^T e^{i \frac{2 \pi (n - n)}{T} t} dt = \int_0^T 1 dt = T
$$

Esta base ortogonal se puede utilizar para representar casi cualquier función periódica en el espacio $L^2_{\left(0, T\right)}$ como una serie de Fourier:
$$
f(t) = \sum_{n=-\infty}^{\infty} c_n  e^{i \frac{2 \pi n}{T} t}
$$
donde los coeficientes $c_n$ se obtienen como:
$$
c_n = \frac{1}{T} \int_0^T f(t) e^{-i \frac{2 \pi n}{T} t} dt \quad \text{para } n \in \mathbb{Z}
$$
Esta fórmula nos da los coeficientes que representan la amplitud y fase de cada componente armónica en la señal periódica.

El estudio de la convergencia y los tipos de funciones que son posibles representar queda fuera de este curso, pero es importante mencionar que la serie de Fourier converge a la función original en casi todos los puntos, excepto en los puntos de discontinuidad, donde converge al promedio de los límites laterales. Para profundizar, capítulo 3 de Señales y Sistemas de Openheim y también [Wikipedia](https://en.wikipedia.org/wiki/Convergence_of_Fourier_series).


### Forma seno-coseno

La serie de Fourier también se puede expresar en términos de funciones seno y coseno. En este caso, la serie de Fourier se escribe como:

$$
f(t) = a_0 + \sum_{n=1}^{\infty} \left( a_n \cos\left(\frac{2 \pi n}{T} t\right) + b_n \sin\left(\frac{2 \pi n}{T} t\right) \right)
$$

donde los coeficientes $a_n$ y $b_n$ se obtienen como:

$$
\begin{align*}
a_0 &= \frac{1}{T} \int_0^T f(t) dt \\
a_n &= \frac{2}{T} \int_0^T f(t) \cos\left(\frac{2 \pi n}{T} t\right) dt &\, \text{para } n \geq 1 \\
b_n &= \frac{2}{T} \int_0^T f(t) \sin\left(\frac{2 \pi n}{T} t\right) dt &\, \text{para } n \geq 1
\end{align*}
$$

### Forma amplitud-fase

La serie de Fourier también se puede expresar en términos de amplitud y fase. En este caso, la serie de Fourier se escribe como:

$$
f(t) = A_0 + \sum_{n=1}^{\infty} A_n \cos\left(\frac{2 \pi n}{T} t - \phi_n\right)
$$
donde $A_n$ es la amplitud y $\phi_n$ es la fase de cada componente armónica. Los coeficientes $A_n$ y $\phi_n$ se obtienen a partir de los coeficientes $a_n$ y $b_n$ como:

$$
\begin{align*}
A_n &= \sqrt{a_n^2 + b_n^2} = \left|c_n\right| \\
\phi_n &= \tan^{-1}\left(\frac{b_n}{a_n}\right) = \text{arg}(c_n)
\end{align*}
$$



## Obtener componentes de una señal

Imaginemos la siguiente señal hecha de tres sinusoides con distintas frecuencias y amplitudes:

$$
y(t) = 2 \sin(2\pi t) + 3 \sin(2\pi 2 t) + 1.5 \sin(2\pi 3 t)
$$

Podríamos preguntarnos si hay alguna forma para extraer los coeficientes de cada senoide teniendo solamente la señal $y(t)$. Dado que estamos en un caso de una señal periódica, podemos hacer el producto interno entre la señal y cada senoide e ir obteniendo los coeficientes de cada senoide.

$$
\langle y(t), \sin(2\pi n t) \rangle = 2 \int_0^{1} y(t) \sin(2\pi n t) dt
$$

### Señal diente de sierra

La señal diente de sierra es una señal periódica que se puede representar como una serie de Fourier. La forma más común de la señal diente de sierra es:

$$
f(t) = \frac{t}{T} \quad  -T/2 \leq t < T/2
$$

donde $T$ es el periodo de la señal. Los coeficientes de la serie de Fourier para la señal diente de sierra se pueden calcular integrando la función en el intervalo $[0, T]$:
$$
\begin{align*}
a_0 &= \frac{1}{T} \int_{-T/2}^{T/2} f(t) dt = \frac{1}{T} \int_{-T/2}^{T/2} \frac{t}{T} dt = 0 \\
a_n &= \frac{2}{T} \int_{-T/2}^{T/2} f(t) \cos\left(\frac{2 \pi n}{T} t\right) dt = \frac{2}{T^2} \int_{-T/2}^{T/2} t \cos\left(\frac{2 \pi n}{T} t\right) dt = 0 \\
b_n &= \frac{2}{T} \int_{-T/2}^{T/2} f(t) \sin\left(\frac{2 \pi n}{T} t\right) dt = \frac{2}{T^2} \int_{-T/2}^{T/2} t \sin\left(\frac{2 \pi n}{T} t\right) dt = \frac{\sin{2 \pi n}}{n^2 \pi^2} - \frac{2 \cos{2 \pi n}}{\pi n} = \left(-1\right)^{n} \frac{2}{\pi n}
\end{align*}
$$

Por lo tanto, la serie de Fourier para la señal diente de sierra es:

$$
f(t) = \sum_{n=1}^{\infty} \left(-1\right)^{n} \frac{2 \sin{\frac{2 \pi n}{T} t }}{\pi n}
$$

## Relación de Parseval

La relación de Parseval establece que la energía total de una señal en el dominio del tiempo es igual a la energía total de la señal en el dominio de la frecuencia. Matemáticamente, se expresa como:

$$
\frac{1}{T} \int_T |f(t)|^2 dt = \sum_{n=-\infty}^{\infty} |c_n|^2
$$

donde $c_n$ son los coeficientes de la serie de Fourier. Esta relación es útil para analizar la energía de una señal en el dominio de la frecuencia.
