# Historia

**Siglo VI a.C.: Pitágoras y la Armonía Musical**  
Relaciones ente longitudes de cuerdas e intervalos musicales. Intervalos simples (como 1:2 o 2:3) producen consonancias.

**1636: Mersenne y las Leyes de las Cuerdas**  
Inicio de la acústica matemática. Relación entre la frecuencia de una cuerda con su longitud, tensión y densidad.

**1807–1822: Fourier y el Análisis Armónico**  
Representación de señales periódicas como suma de senoidales mediante series de Fourier.

**1857: Scott de Martinville y el Fonoautógrafo**  
Édouard-Léon Scott de Martinville creó el fonoautógrafo, el primer dispositivo capaz de registrar visualmente ondas sonoras. Este invento marcó el inicio de la grabación de señales acústicas.

**1863: Helmholtz y la Acústica Moderna**  
Hermann von Helmholtz publicó *Die Lehre von den Tonempfindungen*, explicando la descomposición del sonido en frecuencias puras mediante resonadores. Su obra estableció las bases de la acústica fisiológica y musical.

**1876: Bell y el Teléfono**  
Alexander Graham Bell inventó el teléfono, transmitiendo la voz mediante señales eléctricas. Este avance revolucionó las telecomunicaciones y sentó las bases para la comunicación de audio.

**1877: Edison y el Fonógrafo**  
Thomas Edison desarrolló el fonógrafo, permitiendo grabar y reproducir sonidos. Su invención abrió la era de la preservación y difusión del sonido.

**1888: Hertz y las Ondas Electromagnéticas**  
Heinrich Hertz demostró experimentalmente la existencia de ondas de radio. Sus experimentos confirmaron que las señales electromagnéticas pueden propagarse a través del aire.

**1895: Marconi y la Radio**  
Guglielmo Marconi logró la primera transmisión inalámbrica de señales, extendiendo la comunicación a larga distancia. Su trabajo popularizó la radiocomunicación a nivel global.

**1897: Cahill y el Telharmonium**  
Thaddeus Cahill inventó el Telharmonium, un órgano eléctrico que sintetizaba sonidos mediante ruedas dentadas. Este dispositivo fue un precursor de la música electrónica y la síntesis sonora.

**1906: De Forest y el Audión**  
Lee de Forest introdujo el Audión, un tubo de vacío que amplifica señales eléctricas débiles. Este invento permitió mejorar la transmisión y recepción en sistemas de radio y telecomunicaciones.

**1910: Campbell y los Filtros Electrónicos**  
George A. Campbell diseñó los primeros filtros electrónicos para controlar bandas de frecuencia en telecomunicaciones. Sus aportes facilitaron el manejo eficaz del espectro en sistemas de voz.

**1920: El Theremin de Lev Termen**  
Lev Termen (Léon Theremin) desarrolló el Theremin, uno de los primeros instrumentos musicales electrónicos operados sin contacto físico. Este aparato se convirtió en un ícono de la música electrónica y la síntesis de sonido.

**1927: Yule y el Modelado Estadístico**  
G. Udny Yule introdujo modelos autorregresivos para analizar series temporales en fenómenos naturales. Su trabajo sentó las bases del modelado estadístico de señales.

**1928: Nyquist y el Teorema del Muestreo**  
Harold Nyquist enunció el teorema que define la frecuencia mínima necesaria para muestrear señales de banda limitada sin pérdida de información. Este principio es fundamental para la digitalización de señales.

**1930: Filtro Butterworth**  
Stephen Butterworth desarrolló un filtro analógico con respuesta de amplitud plana en la banda de paso. Este diseño se convirtió en un estándar en la ingeniería de filtros electrónicos.

**1933: Neyman-Pearson y la Detección de Señales**  
Jerzy Neyman y Egon Pearson formularon una teoría óptima para detectar señales en medio de ruido. Su enfoque estadístico es crucial para distinguir información útil en contextos ruidosos.

**1937: Reeves y la Modulación Digital (PCM)**  
Alec Reeves propuso la modulación por impulsos codificados (PCM) para transmitir voz digitalmente. Aunque la tecnología de la época era limitada, su concepto anticipó la era digital.

**1940s: Radar y Cibernética**  
Durante la Segunda Guerra Mundial se desarrollaron sistemas de radar y filtros adaptados para detectar señales débiles. Estos avances se integraron en la cibernética y el control automático de sistemas.

**1948: Shannon y Wiener**  
Claude Shannon publicó la teoría de la información, estableciendo límites teóricos para la comunicación digital. Al mismo tiempo, Norbert Wiener introdujo la cibernética y el filtro de Wiener para la optimización del filtrado de ruido.

**1960: Filtro de Kalman**  
Rudolf Kalman presentó el filtro de Kalman, un método recursivo para estimar el estado de sistemas dinámicos a partir de observaciones ruidosas. Este algoritmo es vital en aplicaciones de seguimiento y navegación en tiempo real.

**1962: PCM T1 y la Telefonía Digital**  
El sistema T1 implementado en Bell Labs marcó la primera transmisión de voz digital mediante PCM. Este avance revolucionó la capacidad y calidad de las telecomunicaciones modernas.

**1963: Teoría del Caos de Lorenz**  
Edward Lorenz descubrió que sistemas deterministas podían comportarse de manera caótica, donde pequeñas variaciones en condiciones iniciales generan grandes diferencias. Este hallazgo introdujo la teoría del caos en el análisis de señales complejas.

**1965: FFT de Cooley-Tukey**  
James Cooley y John Tukey desarrollaron la Transformada Rápida de Fourier (FFT), reduciendo drásticamente el tiempo de cálculo para el análisis espectral. La FFT se convirtió en una herramienta esencial para el procesamiento digital de señales.

**1966: LPC y HMM**  
Itakura y Saito introdujeron la codificación predictiva lineal (LPC) para modelar la voz de manera compacta. Ese mismo año, Leonard Baum formalizó los Modelos Ocultos de Markov (HMM), fundamentales para el reconocimiento de patrones en señales temporales.

**1971: Reconocimiento del Habla y DTW**  
El programa DARPA impulsó el desarrollo de sistemas de reconocimiento del habla, incluyendo algoritmos como Dynamic Time Warping (DTW) para alinear señales. Estos avances establecieron las bases para sistemas de dictado y reconocimiento automático.

**1978: DSP en el Speak & Spell**  
Texas Instruments lanzó el Speak & Spell, que incorporaba el primer chip dedicado a procesamiento digital de señales (DSP) para voz. Este dispositivo demostró el potencial del procesamiento digital en aplicaciones de consumo.

**1980: MFCC para la Voz**  
El método de coeficientes cepstrales en la escala mel (MFCC) se desarrolló para extraer características robustas de la señal de voz. Este enfoque es ahora un estándar en sistemas de reconocimiento automático del habla.

**Década de 1980: Modelos Estadísticos y de Lenguaje**  
La integración de HMM con modelos de lenguaje n-gram permitió avances significativos en el reconocimiento del habla continuo. Investigadores como Fred Jelinek en IBM consolidaron sistemas estadísticos para telecomunicaciones.

**1981: Takens y el Embedding**  
Floris Takens presentó el teorema de inmersión, que permite reconstruir el espacio de estados de un sistema dinámico a partir de series temporales. Esta técnica es crucial para analizar comportamientos caóticos en señales.

**2010s: Redes Neuronales Profundas**  
La aplicación de redes neuronales profundas transformó el reconocimiento del habla y el modelado de lenguaje. Estos métodos aprendieron características directamente de grandes volúmenes de datos, superando enfoques tradicionales.

**2020s: Aprendizaje Automático en Procesamiento de Señales**  
El procesamiento de señales se sigue revolucionando con modelos de aprendizaje automático y técnicas generativas. Estas innovaciones mejoran la calidad en audio, imagen y otros dominios al combinar DSP y reconocimiento de patrones.

1980 (MFCC):
Steven Davis y Paul Mermelstein introdujeron los coeficientes cepstrales de frecuencia mel (MFCC) para extraer características de la voz de forma eficiente. Los MFCC se establecieron como el estándar en sistemas de reconocimiento automático del habla.
