# Transformada Discreta de Fourier

La fórmula de la tranformada de fourier de una señal $x(t)$ es: $$X(f) := \int_{-\infty}^{\infty} x(t)\ e^{-j 2\pi f t}\,dt, $$

La versión discreta es una suma sobre el total de los datos de una señal $x_n$ de largo $N$:

$$X[k] = \sum_{n=0}^{N-1} x[n] e^{- j \frac{2 \pi}{N} k n} \quad \quad k = 0, \dots, N-1$$


La transformada inversa es:

$$x[n] = \frac{1}{N} \sum_{k=0}^{N-1} X[k] e^{j \frac{2 \pi}{N} k n} \quad \quad n = 0, \dots, N-1$$

Esta transformada es posible ya que para un espacio vectorial de dimensión $R^N$ se puede definir una base ortonormal. En este caso, la base es la exponencial compleja $e^{j \frac{2 \pi}{N} k n}$, y el espacio es el de las señales discretas de largo $N$. Visualicemos la base para un $N=10$

```python
import numpy as np
import matplotlib.pyplot as plt

N = 10
# Base exponencial para N

n, k = np.meshgrid(np.arange(N), np.arange(N))
V = np.exp(1j*np.pi*n*k/N)

# subplots 2,1 plot real part first, then imag of V

fig, ax = plt.subplots(1, 2, figsize=(12, 6))
offset = np.arange(0, N)*2
ax[0].plot(np.real(V)-offset, 'k')
ax[0].set_title('Parte real base')
ax[0].set_yticks(np.arange(0, -N, -1)*2)
ax[0].set_yticklabels(np.arange(0, N, 1))
ax[0].set_xlabel('n')
ax[0].set_ylabel('k')

ax[1].plot(np.imag(V)- offset, 'k')
ax[1].set_title('Parte imaginaria base')
ax[1].set_yticks(np.arange(0, -N, -1)*2)
ax[1].set_yticklabels(np.arange(0, N, 1))
ax[1].set_xlabel('n')
ax[1].set_ylabel('k')
plt.tight_layout()

plt.show()
```

## Frecuencias de análisis

Los elementos de la base son exponenciales complejas a frecuencias específicas, todas multiplos de una frecuencia fundamental $\displaystyle  f_0=\frac{1}{N}$, $\displaystyle  f_k =\frac{k}{N}$.

También podemos asignarle al análisis unidades en segundos multiplicando y dividiendo por SR al argumento de la exponencial. El tiempo queda $t=n/SR$ y la frecuencia $f_k$ en Hz, $\displaystyle f_k = \frac{k SR}{N} = \frac{k}{T}$, donde $T$ es el tiempo total de la señal.
