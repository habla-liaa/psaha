# Transformada de Fourier

La transformada de Fourier es la extensión de la serie de Fourier a funciones no periódicas. La transformada de Fourier se define como:

$$
\mathcal{F}\{x(t)\} = X(\omega) = \frac{1}{2\pi} \int_{-\infty}^{\infty} x(t) e^{-j\omega t} dt
$$

donde $x(t)$ es la función en el dominio del tiempo, $X(\omega)$ es la función en el dominio de la frecuencia, y es $\omega$ la frecuencia angular.
