# Energía y Potencia de una señal


Dada una señal $x(t)$ en el caso continuo, y $x[n]$ en el caso discrteo. La potencia instantánea está definida como:

$$
\begin{align}
P(t) = \left| x(t)\right|^2 & \,& P[n] = \left| x[n]\right|^2
\end{align}
$$

Es decir, si la señal toma valores reales, la potencia es elevar al cuadrado a la señal.


En terminos físicos, la potencia es energía por unidad de tiempo, por lo que si sumamos la potencia instantanea entre el tiempmo $T_1$ y el tiempo $T_2$ deberíamos obtener la energía total de ese intervalo de tiempo.

$$
\begin{align}
E_{T_1,T_2} = \int_{T_1}^{T_2} \left| x(t)\right|^2 dt & \,& E_{n_1,n_2} = \sum_{n = n_1}^{n_2}\left| x[n]\right|^2
\end{align}
$$

En el caso continuo esta dado por una integral. En el caso discreto está dado por una sumatoria de la señal al cuadrado.


Si a esta energía $E(t)$ las dividimos por el total del tiempo del intervalo $T_2-T_1$, obtenemos de nuevo algo con unidades de potencia. Pero esta vez no es la potencia instantáneas, sino la potencia promedio, ya que hay una suma y una división por el intervalo.

$$
\begin{align}
P_{T_1,T_2} = \int_{T_1}^{T_2} \frac{\left| x(t)\right|^2 dt}{T_2-T_1} & \,& P_{n_1,n_2} = \sum_{n =n_1}^{n_2}\frac{\left| x[n]\right|^2}{n_2-n_1+1}
\end{align}
$$

Notemos que en el caso discreto el denominador tiene un 1 agregado para contemplar el caso cuando $n_2-n_1 = 0$


Finalmente definimos el valor RMS de una señal $x(t)$ como el valor que tiene que tener una señal constante para tener la misma potencia que la señal original.

$$
\begin{align}
X_{RMS} = \sqrt{\int_{T_1}^{T_2} \frac{\left| x(t)\right|^2 dt}{T_2-T_1}} & \,& X_{RMS} = \sqrt{\sum_{n = n_1}^{n_2}\frac{\left| x[n]\right|^2}{n_2-n_1+1}}
\end{align}
$$

El valor RMS sirve como valor eficaz de la señal y tiene las mismas unidades que la señal.

Realicemos el cálculo para el caso de una señal sinusoide con frecuencia $f$ $x(t) = A\sin(2\pi f t)$:


$$ X_{RMS} = \sqrt{\int_{0}^{T} \frac{ (A\sin(2\pi f t))^2 dt}{T}} $$


Donde $T$ es el período de la señal $1/f$. El resultado de la integral es $\frac{A^2}{2}T$. Falta dividir por $T$ y aplicar la raiz. Finalmente queda:


$$ X_{RMS} = \sqrt{\frac{A^2}{2}T/T} = \frac{1}{\sqrt{2}}A = 0.707 A$$


Estas magnitudes pueden expresarse en escala de decibeles a través de la siguiente fórmula genérica para una magnitud L:


$$ L_{dB} = 10 \log_{10}\left(\frac{L}{L_{ref}}\right) $$


Donde $L_{ref}$ es un valore de referencia para el cual el valor en decibeles $L_{dB}$ vale 0.


Que en el caso de la potencia de una señal sería:


$$ P_{dB} = 10 \log_{10}\left(\frac{P}{P_{ref}}\right) $$


$$ P_{dB} = 10 \log_{10}\left(\frac{X_{RMS}^2}{X^2_{ref}}\right) = 20 \log_{10}\left(\frac{X_{RMS}}{X_{ref}}\right) $$


Para el caso de una sinusoide quedaría: $P_{dB} = 20 \log_{10}\left(\frac{A}{A_{ref}}\right) $


Uno de los casos más comunes es la intensidad del sonido medida en $dB SPL$, (SPL = sound pressure level) cuya fórmula es:


$$ 
L_p=10 \log_{10}\left(\frac{{p_{\mathrm{{RMS}}}}^2}{{p_{\mathrm{ref}}}^2}\right) =20 \log_{10}\left(\frac{p_{\mathrm{RMS}}}{p_{\mathrm{ref}}}\right) dB
$$


El valor para $p_{ref}$ suele tomarse como la amplitud de un tono puro de $1kHz$ en el umbral de la audición, aproximadamente $20$ micropascales, $p_{ref}=20\mu Pa$
