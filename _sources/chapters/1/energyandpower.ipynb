# Energía y Potencia de una señal

Primero escribamos las expresiones de media y varianza para señales continuas y discretas. Podemos definir la media de una señal como:

$$
\begin{align}
\mu = \lim_{T\to\infty}\frac{1}{T}\int_{-T/2}^{T/2}x(t)dt & \,& \mu = \lim_{N\to\infty}\frac{1}{N}\sum_{n=1}^{N}x[n]
\end{align}
$$

Y la varianza de la señal como:

$$
\begin{align}
\sigma^2 = \lim_{T\to\infty}\frac{1}{T}\int_{-T/2}^{T/2}(x(t)-\mu)^2dt & \,& \sigma^2 = \lim_{N\to\infty}\frac{1}{N}\sum_{n=1}^{N}(x[n]-\mu)^2
\end{align}
$$  

Ahora vamos a definir la potencia instantánea como:

$$
\begin{align}
P(t) = \left| x(t)\right|^2 & \,& P[n] = \left| x[n]\right|^2
\end{align}
$$

Es decir, si la señal toma valores reales, la potencia es elevar al cuadrado a la señal.


En terminos físicos, la potencia es energía por unidad de tiempo, por lo que si sumamos la potencia instantanea entre el tiempmo $T_1$ y el tiempo $T_2$ deberíamos obtener la energía total de ese intervalo de tiempo.

$$
\begin{align}
E_{T_1,T_2} = \int_{T_1}^{T_2} \left| x(t)\right|^2 dt & \,& E_{n_1,n_2} = \sum_{n = n_1}^{n_2}\left| x[n]\right|^2
\end{align}
$$

Si a esta energía $E(t)$ las dividimos por el total del tiempo del intervalo $T_2-T_1$, obtenemos nuevamente un valor con unidades de potencia. Esta vez no es la potencia instantánea, sino la potencia promedio.

$$
\begin{align}
P_{T_1,T_2} = \int_{T_1}^{T_2} \frac{\left| x(t)\right|^2 dt}{T_2-T_1} & \,& P_{n_1,n_2} = \sum_{n =n_1}^{n_2}\frac{\left| x[n]\right|^2}{n_2-n_1+1}
\end{align}
$$

Notemos que en el caso discreto el denominador tiene un 1 agregado para contemplar el caso cuando $n_2-n_1 = 0$.

Si la señal tiene media nula la definición de potencia es equivalente a calcular la varianza de la señal, y si la media no es nula, la potencia se puede escribir como la varianza más la media al cuadrado:

$$
\begin{align}
P = \sigma^2 + \mu^2
\end{align}
$$

Finalmente definimos el valor RMS de una señal $x(t)$ como el valor que tiene que tener una señal constante para tener la misma potencia que la señal original.

$$
\begin{align}
X_{RMS} = \sqrt{\int_{T_1}^{T_2} \frac{\left| x(t)\right|^2 dt}{T_2-T_1}} & \,& X_{RMS} = \sqrt{\sum_{n = n_1}^{n_2}\frac{\left| x[n]\right|^2}{n_2-n_1+1}}
\end{align}
$$

El valor RMS sirve como valor eficaz de la señal y tiene las mismas unidades que la señal.

Realicemos el cálculo para el caso de una señal sinusoide con frecuencia $f$ $x(t) = A\sin(2\pi f t)$:


$$ X_{RMS} = \sqrt{\int_{0}^{T} \frac{ (A\sin(2\pi f t))^2 dt}{T}} $$


Donde $T$ es el período de la señal $1/f$. El resultado de la integral es $\frac{A^2}{2}T$. Falta dividir por $T$ y aplicar la raiz. Finalmente queda:


$$ X_{RMS} = \sqrt{\frac{A^2}{2}T/T} = \frac{1}{\sqrt{2}}A = 0.707 A$$

## Densidad de una señal

Dada una señal $x(t)$ consideramos que la densidad en función de $x$ depende de cuanto tiempo pase la señal en ese valor.

$$
p(x) = \frac{dt}{T dx}
$$

Con $T$ el tiempo total de la señal. La probabilidad de que la señal tome un valor entre $x$ y $x+dx$ es $P(x)dx$. 

Si conocemos la densidad de probabilidad de una señal, podemos calcular la media y la varianza de la señal con las formulas de los momentos. La esperanza de una señal es:

$$
\begin{align}
E[x(t)] = \int_{-\infty}^{\infty} x p(x) dx
\end{align}
$$

Donde $p(x)$ es la función de densidad de probabilidad de la señal. La varianza de la señal es:

$$
\begin{align}
Var[x(t)] = \int_{-\infty}^{\infty} (x-E[x(t)])^2 p(x) dx
\end{align}
$$

Con esta fórmula podemos calcular la potencia de una señal si conocemos la densidad de probabilidad de la señal. La potencia de una señal es:

$$
\begin{align}
P = Var[x(t)] + E[x(t)]^2 = \int_{-\infty}^{\infty} (x - \mu)^2 p(x) dx + \mu^2
\end{align}
$$

## Decibeles

La energía y la potencia pueden expresarse en escala de decibeles a través de la siguiente fórmula genérica para una magnitud L:


$$ L_{dB} = 10 \log_{10}\left(\frac{L}{L_{ref}}\right) $$


Donde $L_{ref}$ es un valore de referencia para el cual el valor en decibeles $L_{dB}$ vale 0.


Que en el caso de la potencia de una señal sería:


$$ P_{dB} = 10 \log_{10}\left(\frac{P}{P_{ref}}\right) $$


$$ P_{dB} = 10 \log_{10}\left(\frac{X_{RMS}^2}{X^2_{ref}}\right) = 20 \log_{10}\left(\frac{X_{RMS}}{X_{ref}}\right) $$


Para el caso de una sinusoide quedaría: $P_{dB} = 20 \log_{10}\left(\frac{A}{A_{ref}}\right) $


Uno de los casos más comunes es la intensidad del sonido medida en $dB_{SPL}$, (SPL = sound pressure level) cuya fórmula es:


$$ 
L_p=10 \log_{10}\left(\frac{{p_{\mathrm{{RMS}}}}^2}{{p_{\mathrm{ref}}}^2}\right) =20 \log_{10}\left(\frac{p_{\mathrm{RMS}}}{p_{\mathrm{ref}}}\right) dB
$$


El valor para $p_{ref}$ suele tomarse como la amplitud de un tono puro de $1kHz$ en el umbral de la audición, aproximadamente $20$ micropascales, $p_{ref}=20\mu Pa$
