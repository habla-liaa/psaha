# Ejercicios Capítulo 3


## Ejercicios Habla (1)

### Ejercicios 1.1

Queremos computar el pitch de una señal de habla usando la autocorrelación. Para esto computar un autocorrelograma de una señal de habla, es decir la autocorrelación para ventanas solapadas. Luego graficar como una imagen la autocorrelación. Para cada ventana encontrar un pico en la autocorrelación en el rango de períodos esperados para el habla utilizada. Establecer un criterio para descartar ventanas donde no hay pitch.
